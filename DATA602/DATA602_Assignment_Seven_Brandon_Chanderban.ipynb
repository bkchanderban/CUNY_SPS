# Data Exploration



```
# Import the required library.
import pandas as pd

# Load the Video Game Sales with Ratings dataset from GitHub.
url = "https://raw.githubusercontent.com/bkchanderban/CUNY_SPS/refs/heads/main/DATA602/Video_Games_Sales_as_at_22_Dec_2016.csv"
df = pd.read_csv(url)

# Display the dimensions and the first five rows of the dataset.
print(f"The dataset's dimensions are: {df.shape}\n")
display(df.head())
print()

# Examine the dataset's columns, data types, and non-null values.
df.info()
```

```
# Generate the summary statistics for the numeric columns.
print("Summary Statistics:")
display(df.describe())

# Calculate the mean and median for the numerical columns.
print("\nMean Values:")
display(df.mean(numeric_only=True))

print("\nMedian Values:")
display(df.median(numeric_only=True))

# Calculate the first, second and third quartiles.
print("\nQuartiles:")
display(df.quantile([0.25, 0.50, 0.75], numeric_only=True))

# Identify missing values within the dataset.
print("\nMissing Values by Column:")
display(df.isnull().sum())

# Calculate the total number of missing values.
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

##**Interpretation**

The initial interpretation revealed that the dataset contains a total of 16,719 records across 16 columns, with recorded release years ranging from 1980 to 2020. With respect to global sales, the mean was approximately 0.53 million units, whilst the median was considerably lower at 0.17 million units. This suggests that the distribution is positively skewed, with a relatively small number of highly successful games contributing disproportionately to the total global sales.

Moreover, the dataset contains 46,716 missing values/cells across a number of columns, especially those associated with critic and user ratings, review counts, and developer information. Furthermore, the User_Score column is also stored as an object rather than a numerical data type, indicating that further examination and potential data type conversion will be necessary. These findings thus highlight several opportunities for data cleaning and manipulation in the subsequent section.

# Data Wrangling





###**1. Creating the Subset and Renaming Columns**

```
# Create a random subset of 1000 records from the original video games dataset.
df_subset = df.sample(n=1000, random_state=6767).copy()

# Display the dimensions and first five rows of the subset.
print(f"The subset's dimensions are: {df_subset.shape}\n")
display(df_subset.head())

# Renaming multiple columns for the purpose of improved readability.
df_subset = df_subset.rename(columns={
    "Name": "Game_Title",
    "NA_Sales": "North_American_Sales_Millions",
    "EU_Sales": "European_Sales_Millions",
    "JP_Sales": "Japanese_Sales_Millions",
    "Other_Sales": "Other_Regions_Sales_Millions"
})

# Display the updated column names.
print("\nUpdated Column Names:")
display(df_subset.columns.to_list())
```

###**2. Examining and Correcting Data Types**

```
# Examine the data types of the subset before conducting conversion.
print("Data Types Before Conversion:")
print(df_subset[["Year_of_Release", "User_Score"]].dtypes)

# Convert the release year to a nullable integer data type.
df_subset["Year_of_Release"] = df_subset["Year_of_Release"].astype("Int64")

# Convert the user scores to numerical values, treating invalid entries as missing.
df_subset["User_Score"] = pd.to_numeric(df_subset["User_Score"], errors="coerce")

# Display the updated data types.
print("\nData Types After Conversion:")
print(df_subset[["Year_of_Release", "User_Score"]].dtypes)
```

###**3. Handling Missing Data**

```
# Examine missing values before cleaning.
print("Missing Values Before Cleaning:")
display(df_subset.isnull().sum())

# Replace the missing publisher and developer names with "Unknown".
df_subset["Publisher"] = df_subset["Publisher"].fillna("Unknown")
df_subset["Developer"] = df_subset["Developer"].fillna("Unknown")

# Replacing missing game ratings with "Unrated".
df_subset["Rating"] = df_subset["Rating"].fillna("Unrated")

# Check missing values after cleaning.
print("\nMissing Values After Cleaning:")
display(df_subset.isnull().sum())
```

The initial examination of missing values revealed that several columns contained incomplete information, particularly those associated with critic and user scores, developer details, and game ratings. To address this, missing publisher and developer names were replaced with "Unknown", whilst missing game ratings were classified as "Unrated". However, missing numerical scores were retained rather than replaced with arbitrary values, as doing so could potentially distort the results of subsequent analyses. This approach allowed for the cleaning of selected categorical values whilst preserving the integrity of the numerical data.

###**4. Creating New Columns Based on Existing Sales Data**

```
# Calculate the percentage of global sales generated in North America.
df_subset["North_American_Sales_Percentage"] = (
    df_subset["North_American_Sales_Millions"] / df_subset["Global_Sales"]
) * 100

# Calculate the percentage of global sales generated in Europe.
df_subset["European_Sales_Percentage"] = (
    df_subset["European_Sales_Millions"] / df_subset["Global_Sales"]
) * 100

# Display the updated dimensions.
print(f"Updated Subset Dimensions: {df_subset.shape}\n")

# Display the new columns alongside the original sales figures.
display(df_subset[[
    "Game_Title",
    "Global_Sales",
    "North_American_Sales_Percentage",
    "European_Sales_Percentage"
]].head())
```

Two additional columns were created to calculate the percentage of global sales attributed to North America and Europe, respectively. These calculations provided a clearer understanding of how sales are distributed across different geographical markets. For instance, whilst *Playboy: The Mansion* generated 50% of its global sales in North America, *Total War: WARHAMMER* generated approximately 92.31% of its global sales in Europe. This demonstrates how regional sales contributions can vary considerably between individual games.

###**5. Dropping Columns from the Dataset**

```
# Drop the critic and user review count columns.
df_subset = df_subset.drop(columns=["Critic_Count", "User_Count"])

# Display the updated dimensions and remaining columns.
print(f"Updated Subset Dimensions: {df_subset.shape}")

print("\nRemaining Columns:")
display(df_subset.columns.to_list())
```

###**6. Dropping Rows from the Dataset**

```
# Examine the number of rows before removing missing release years.
print(f"Number of Rows Before Removing Missing Release Years: {len(df_subset)}")

# Drop rows where the release year is missing.
df_subset = df_subset.dropna(subset=["Year_of_Release"])

# Display the updated number of rows and dimensions.
print(f"Number of Rows After Cleaning: {len(df_subset)}")
print(f"Updated Subset Dimensions: {df_subset.shape}")

# Confirm that no missing release years remain.
print(f"Missing Release Years: {df_subset['Year_of_Release'].isnull().sum()}")
```

Following the removal of records with missing release years, the dataset was reduced from 1,000 to 981 observations, representing the exclusion of 19 records. This was done to ensure that subsequent analyses involving release years would not be affected by incomplete information. Rather than removing all records containing missing values, the cleaning process focussed specifically on the release year, thereby preserving other observations that may still provide useful information for the analysis.

###**7. Sorting the Dataset Based on Multiple Variables**

```
# Sort the dataset by genre (ascending) and global sales (descending).
df_subset = df_subset.sort_values(by=["Genre", "Global_Sales"], ascending=[True, False])

# Display the first eight rows of the sorted dataset.
display(df_subset[[
    "Game_Title",
    "Genre",
    "Platform",
    "Global_Sales"
]].head(8))
```

###**8. Filtering the Dataset Based on a Condition**

```
# Filter games released from 2010 onward with at least 1 million in global sales.
df_filtered = df_subset[
    (df_subset["Year_of_Release"] >= 2010) & (df_subset["Global_Sales"] >= 1.0)
]

# Display the number of records meeting both conditions.
print(f"The number of records meeting both conditions is: {len(df_filtered)}\n")

# Display the first ten records of the filtered dataset.
display(df_filtered[[
    "Game_Title",
    "Platform",
    "Year_of_Release",
    "Genre",
    "Global_Sales"
]].head(10))
```

The filtering process identified 44 video games released from 2010 onward that generated at least one million units in global sales. By applying both conditions simultaneously, the analysis isolated a smaller group of relatively recent, commercially successful titles within the sampled dataset. Moreover, the filtered results were stored in a separate DataFrame, allowing the original working subset to be retained for subsequent analysis.

###**9. Converting String Values to Uppercase or Lowercase**

```
# Display the genre values before conversion.
print("Genre Values Before Conversion:")
display(df_subset["Genre"].unique())

# Convert all genre values to uppercase.
df_subset["Genre"] = df_subset["Genre"].str.upper()

# Display the updated genre values.
print("\nGenre Values After Conversion:")
display(df_subset["Genre"].unique())
```

###**10. Checking for Numeric Values in a Column**

```
# Check the data type of the User_Score column.
print(f"The User_Score column's data type is: {df_subset['User_Score'].dtype}")

# Identify the number of valid numeric values.
numeric_count = df_subset["User_Score"].notna().sum()

# Identify the number of missing values.
missing_count = df_subset["User_Score"].isna().sum()

# Display the results.
print(f"Number of Valid Numeric Values: {numeric_count}")
print(f"Number of Missing Values: {missing_count}")

# Confirm whether the column contains numerical data.
print(f"Is the attribute User_Score numeric? {pd.api.types.is_numeric_dtype(df_subset['User_Score'])}")
```

###**11. Grouping by One Column and then Calculating Summary Statistics**

```
# Group the dataset by genre and then calculate the mean, minimum, and maximum global sales.
genre_sales_summary = df_subset.groupby("Genre")["Global_Sales"].agg(
    ["mean", "min", "max"]
)

# Rename the aggregated columns for improved readability.
genre_sales_summary = genre_sales_summary.rename(columns={
    "mean": "Mean_Global_Sales",
    "min": "Minimum_Global_Sales",
    "max": "Maximum_Global_Sales"
})

# Display the aggregated sales statistics by genre.
print("Global Sales Summary by Genre (Millions of Units):")
display(genre_sales_summary.sort_values(by="Mean_Global_Sales", ascending=False))
```

The grouped analysis revealed noticeable differences in global sales across video game genres. Within the sampled dataset, the SHOOTER genre recorded the highest average global sales at approximately 0.983 million units, followed by PLATFORM and ROLE-PLAYING at approximately 0.708 million and 0.705 million units, respectively. Conversely, ADVENTURE recorded the lowest average at approximately 0.113 million units. These findings suggest that sales performance varies considerably across genres, although the results should be interpreted within the context of the sampled observations rather than the wider video game market.

###**12. Grouping by Two Columns and Sorting Aggregated Results**

```
# Group the dataset by genre and platform, calculating the average global sales.
genre_platform_sales = df_subset.groupby(
    ["Genre", "Platform"])["Global_Sales"].agg(["mean", "count"])

# Rename the aggregated columns for improved readability.
genre_platform_sales = genre_platform_sales.rename(columns={
    "mean": "Mean_Global_Sales",
    "count": "Number_of_Games"
})

# Sort the results by genre and descending average global sales within each genre.
genre_platform_sales = genre_platform_sales.sort_values(
    by=["Genre", "Mean_Global_Sales"], ascending=[True, False]
)

# Display the first 20 records of the grouped and sorted dataset.
print("Average Global Sales by Genre and Platform (Millions of Units):")
display(genre_platform_sales.head(20))
```

The analysis further examined global sales by grouping video games according to both genre and platform. Within each genre, platforms were arranged in descending order of average global sales, allowing for a more detailed comparison of sales performance. For instance, within the ACTION genre, the original PlayStation (PS) recorded the highest average global sales at approximately 1.71 million units across 11 games. However, differences in the number of games represented across platforms should be considered when interpreting these averages, particularly where a small number of observations may influence the results.

###**Additional Analysis: Relationship Between Critic Scores and Global Sales**

```
# Create a subset containing games with valid critic scores and global sales.
df_correlation = df_subset.dropna(subset=["Critic_Score", "Global_Sales"])

# Calculate the Pearson correlation between critic scores and global sales.
critic_sales_correlation = df_correlation[
    ["Critic_Score", "Global_Sales"]
].corr(method = "pearson")

# Display the number of observations included in the analysis.
print(f"Number of Games Analyzed: {len(df_correlation)}")

# Display the correlation matrix.
print("\nCorrelation Between Critic Scores and Global Sales:")
display(critic_sales_correlation)

# Display the correlation coefficient separately.
print(
    f"\nPearson Correlation Coefficient: "
    f"{critic_sales_correlation.loc["Critic_Score", "Global_Sales"]:.3f}"
)
```

The additional analysis examined the relationship between critic scores and global sales across 480 games with valid observations. The Pearson correlation coefficient of approximately 0.265 indicated a weak positive relationship, suggesting that games receiving higher critic scores tended to generate somewhat higher global sales. However, the relatively low coefficient suggests that critic ratings alone may not be a strong indicator of commercial performance, with other factors potentially contributing to differences in sales. Moreover, the exclusion of records with missing critic scores should be considered when interpreting these findings. It is also paramount to note that correlation does not necessarily imply causation.
