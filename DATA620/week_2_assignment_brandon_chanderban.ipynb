```python
# Brandon Chanderban
# DATA 620 - Web Analytics
# Week 2 Assignment: Hello Graph World

# Import relevant libraries
import networkx as nx
import matplotlib.pyplot as plt

# Create an empty undirected graph in which connections go both ways.
# For instance, if Heather is connected to Ike, Ike is also connected to Heather.
G = nx.Graph()

# Specify all the nodes (persons within the network) in a nodes list
nodes = [
    "Andre",
    "Beverly",
    "Carol",
    "Diane",
    "Ed",
    "Fernando",
    "Garth",
    "Heather",
    "Ike",
    "Jane",
]

# Add all the nodes within the nodes list to the network in one batch
G.add_nodes_from(nodes)

# Set out all the edges (18) in the network (the relationships existing between
# the persons)
edges = [
    ("Andre", "Carol"),
    ("Andre", "Diane"),
    ("Andre", "Beverly"),
    ("Andre", "Fernando"),
    ("Beverly", "Diane"),
    ("Beverly", "Ed"),
    ("Beverly", "Garth"),
    ("Carol", "Diane"),
    ("Carol", "Fernando"),
    ("Diane", "Ed"),
    ("Diane", "Fernando"),
    ("Diane", "Garth"),
    ("Ed", "Garth"),
    ("Fernando", "Garth"),
    ("Fernando", "Heather"),
    ("Garth", "Heather"),
    ("Heather", "Ike"),
    ("Ike", "Jane"),
]
# Add all the edges within the edges list to the network in one batch
G.add_edges_from(edges)

# Confirm the number of created nodes and edges within the network
print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")

# Specify the positions of the node to reflect the provided visualization
pos = {
    "Andre": (0, 3),
    "Beverly": (0, 1),
    "Carol": (0, 2),
    "Diane": (1, 2),
    "Ed": (1, 0),
    "Fernando": (2, 3),
    "Garth": (2, 1),
    "Heather": (3, 2),
    "Ike": (4, 2),
    "Jane": (5, 2),
}

# Draw the concocted network
nx.draw(
    G,
    pos,
    with_labels = True,
    node_size= 1800,
    node_shape = "s",
    font_size= 8,
    font_color = "white",
    edge_color = "purple",
    width = 2,
    font_weight = "bold"
    )

# Add a title to the visualization
plt.title("Assignment: 'hello, graph world' (Krackhardt Kite)")

# Display the final visualization
plt.show()
```
